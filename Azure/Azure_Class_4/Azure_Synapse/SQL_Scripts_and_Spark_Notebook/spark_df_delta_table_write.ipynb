```python
%%pyspark
from pyspark.sql.functions import col, sum
df = spark.read.load('abfss://synapse-data-store@adlsaccountgds.dfs.core.windows.net/Raw_Data/sales_data.csv', format='csv', header=True, inferSchema=True)

# Define the schema (database)
spark.sql("CREATE DATABASE IF NOT EXISTS salesmart")

# Perform some aggregation (e.g., total sales by year)
aggregated_df = df.groupBy("YEAR_ID").agg(
    sum(col("SALES")).alias("total_sales")
)

# Write the Delta table without specifying a path
aggregated_df.write.format("delta").mode("overwrite").saveAsTable("salesmart.SalesAggregation")

print("Write in Delta Completed !!!!!")
```
